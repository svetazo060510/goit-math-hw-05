# Задача 1. Аналіз рейтингів мобільного додатку

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
def analyze_app_ratings():
    # 1. Початкові дані
    ratings = np.array([1, 2, 3, 4, 5])
    counts = np.array([120, 180, 320, 580, 800])
    total_reviews = counts.sum()

    # 2. Обчислення PMF (Probability Mass Function)
    # P(X = x) = кількість відгуків / загальна кількість
    pmf = counts / total_reviews

    # 3. Обчислення CDF (Cumulative Distribution Function)
    # F(x) = P(X <= x) за допомогою кумулятивної суми
    cdf = np.cumsum(pmf)

    # 4. Вивід таблиці результатів
    df = pd.DataFrame({
        'Оцінка (x)': ratings,
        'PMF (P(X=x))': pmf.round(4),
        'CDF (F(x))': cdf.round(4)
    })
    print("Таблиця розподілу рейтингу:")
    print(df.to_string(index=False))
    print("-" * 30)

    # 5. Числові характеристики
    # Математичне сподівання E[X] = Σ x * P(X=x)
    expected_value = np.sum(ratings * pmf)

    # Дисперсія Var(X) = E[X^2] - (E[X])^2
    expected_value_sq = np.sum((ratings**2) * pmf)
    variance = expected_value_sq - (expected_value**2)

    # Стандартне відхилення σ = sqrt(Var(X))
    std_dev = np.sqrt(variance)

    # 6. Ймовірності для груп
    # Негативна оцінка (1 або 2 зірки) -> P(X <= 2) або CDF(2)
    p_negative = cdf[1] # індекс 1 відповідає оцінці 2

    # Висока оцінка (4 або більше) -> P(X >= 4) = P(X=4) + P(X=5)
    p_high = pmf[3] + pmf[4] # або 1 - cdf[2]

    # 7. Медіана рейтингу
    # Найменше x, для якого F(x) >= 0.5
    median = ratings[np.where(cdf >= 0.5)[0][0]]

    # Вивід результатів
    print(f"Математичне сподівання E[X]: {expected_value:.4f}")
    print(f"Дисперсія Var(X):           {variance:.4f}")
    print(f"Стандартне відхилення σ:    {std_dev:.4f}")
    print(f"Ймовірність негативної оцінки (1-2 star): {p_negative:.4f} ({p_negative*100:.2f}%)")
    print(f"Ймовірність високої оцінки (4-5 star):    {p_high:.4f} ({p_high*100:.2f}%)")
    print(f"Медіана рейтингу:           {median}")
```

```python
if __name__ == "__main__":
    analyze_app_ratings()
```

# Задача 2. Аналіз доставки замовлень

```python
from scipy.stats import binom
import numpy as np
```

```python
def analyze_delivery_reliability():
    # 1. Параметри розподілу
    n = 50        # загальна кількість замовлень
    p = 0.88      # ймовірність вчасної доставки
    q = 1 - p     # ймовірність запізнення

    # 2. Математичне сподівання та стандартне відхилення
    # Для біноміального розподілу: E[X] = n*p, Var(X) = n*p*q
    expected_value = n * p
    variance = n * p * q
    std_dev = np.sqrt(variance)

    # 3. Ймовірність, що всі 50 замовлень доставлять вчасно: P(X = 50)
    p_all_50 = binom.pmf(50, n, p)

    # 4. Ймовірність, що рівно 45 замовлень доставлять вчасно: P(X = 45)
    p_exactly_45 = binom.pmf(45, n, p)

    # 5. Ймовірність, що вчасно доставлять від 42 до 46 замовлень: P(42 <= X <= 46)
    # Обчислюється як F(46) - F(41)
    p_range_42_46 = binom.cdf(46, n, p) - binom.cdf(41, n, p)

    # 6. Ймовірність, що більше 5 замовлень запізняться
    # "Більше 5 запізняться" означає, що вчасно приїде МЕНШЕ ніж 45 (50 - 5 = 45)
    # P(Запізн > 5) = P(Вчасно < 45) = P(X <= 44)
    p_more_than_5_late = binom.cdf(44, n, p)

    # Вивід результатів
    print(f"--- Аналіз біноміального розподілу (n={n}, p={p}) ---")
    print(f"Математичне сподівання E[X]: {expected_value:.2f}")
    print(f"Стандартне відхилення σ:    {std_dev:.4f}")
    print("-" * 45)
    print(f"P(X = 50) [Всі вчасно]:      {p_all_50:.6f}")
    print(f"P(X = 45) [Рівно 45]:        {p_exactly_45:.4f}")
    print(f"P(42 <= X <= 46) [Діапазон]: {p_range_42_46:.4f}")
    print(f"P(Запізняться > 5) [X <= 44]: {p_more_than_5_late:.4f}")
```

```python
if __name__ == "__main__":
    analyze_delivery_reliability()
```

# Задача 3. Час між замовленнями в інтернет-магазині

```python
from scipy.stats import expon
import numpy as np
```

```python
def analyze_order_flow():
    # 1. Параметри розподілу
    # lambda = 0.1 зам/хв, отже scale = 1/lambda = 10 хвилин
    avg_time = 10

    # Створюємо об'єкт розподілу для зручності
    dist = expon(scale=avg_time)

    # 2. Математичне сподівання E[T] та стандартне відхилення sigma
    # Для показникового розподілу E[T] = sigma = 1/lambda
    expected_value = dist.mean()
    std_dev = dist.std()

    # 3. Ймовірність, що замовлення надійде менше ніж за 5 хвилин: P(T < 5)
    p_less_5 = dist.cdf(5)

    # 4. Ймовірність, що доведеться чекати більше 15 хвилин: P(T > 15)
    # Обчислюється як 1 - F(15) або через Survival Function (sf)
    p_more_15 = dist.sf(15)

    # 5. Ймовірність у проміжку від 5 до 15 хвилин: P(5 < T < 15)
    # Обчислюється як F(15) - F(5)
    p_between_5_15 = dist.cdf(15) - dist.cdf(5)

    # 6. Медіанний час очікування
    # Значення t, для якого F(t) = 0.5. Використовуємо Percent Point Function (ppf)
    median_time = dist.ppf(0.5)

    # Вивід результатів
    print(f"--- Аналіз показникового розподілу (scale={avg_time} хв) ---")
    print(f"Математичне сподівання E[T]: {expected_value:.2f} хв")
    print(f"Стандартне відхилення σ:    {std_dev:.2f} хв")
    print("-" * 50)
    print(f"P(T < 5) [Менше 5 хв]:       {p_less_5:.4f} ({p_less_5*100:.2f}%)")
    print(f"P(T > 15) [Більше 15 хв]:    {p_more_15:.4f} ({p_more_15*100:.2f}%)")
    print(f"P(5 < T < 15) [Проміжок]:    {p_between_5_15:.4f} ({p_between_5_15*100:.2f}%)")
    print(f"Медіанний час очікування:    {median_time:.2f} хв")
```

```python
if __name__ == "__main__":
    analyze_order_flow()
```

# Задача 4. Перевірка середнього балу студентів

```python
from scipy import stats
import numpy as np
```

```python
def hypothesis_testing_exam():
    # 1. Початкові дані
    mu_0 = 72          # середнє минулого року (нульова гіпотеза)
    sample_mean = 76.8 # вибіркове середнє
    s = 14.2           # вибіркове стандартне відхилення
    n = 35             # розмір вибірки
    alpha = 0.05       # рівень значущості
    df = n - 1         # ступені свободи

    # 2. Формулювання гіпотез (двосторонній тест)
    # H0: mu = 72 (нова методика не змінила результати)
    # H1: mu != 72 (результати змінилися)

    # 3. Обчислення стандартної похибки (SE)
    se = s / np.sqrt(n)

    # 4. Обчислення t-статистики
    t_stat = (sample_mean - mu_0) / se

    # 5. Обчислення p-value для двостороннього тесту
    # stats.t.cdf(t, df) дає ймовірність зліва.
    # Для двостороннього тесту беремо 2 * (1 - cdf(|t|))
    p_value = 2 * (1 - stats.t.cdf(np.abs(t_stat), df))

    # 6. Побудова 95% довірчого інтервалу
    # Знаходимо критичне значення t для рівня 1 - alpha/2
    t_crit = stats.t.ppf(1 - alpha/2, df)
    margin_of_error = t_crit * se
    ci_lower = sample_mean - margin_of_error
    ci_upper = sample_mean + margin_of_error

    # Вивід результатів
    print(f"--- Результати статистичного тесту ---")
    print(f"Стандартна похибка (SE): {se:.4f}")
    print(f"t-статистика:           {t_stat:.4f}")
    print(f"p-value:                {p_value:.4f}")
    print(f"Критичне значення t:    {t_crit:.4f}")
    print(f"95% довірчий інтервал:  [{ci_lower:.2f}, {ci_upper:.2f}]")
    print("-" * 40)

    # 7. Висновок
    if p_value < alpha:
        conclusion = "Відкидаємо нульову гіпотезу (H0)."
        result = "Результати статистично значущі: нова методика вплинула на середній бал."
    else:
        conclusion = "Немає підстав відхилити нульову гіпотезу (H0)."
        result = "Результати статистично не значущі: вплив нової методики не доведено."

    print(f"Рішення: {conclusion}")
    print(f"Висновок: {result}")
```

```python
if __name__ == "__main__":
    hypothesis_testing_exam()
```
